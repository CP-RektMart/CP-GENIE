```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
```

```python
RETRIEVED_METRICS = ['Precision@1', 'Precision@5', 'Precision@10', 'Mean Reciprocal Rank']
```

```python
# bge-m3, jina-v3, multi-e5-large-ins, nomic-embed-v2, LaBSE
score_bge = [ 0.7954, 0.9563, 0.9795, 0.8662 ]
score_jina = [ 0.6439, 0.9195, 0.9686,  0.7683 ]
score_e5 = [ 0.7435, 0.9495, 0.9700, 0.8305 ]
score_nomic = [ 0.6780, 0.9250, 0.9645, 0.7853 ]
score_labse = [ 0.3793, 0.6371, 0.7299, 0.4885]
```

```python
fig = go.Figure()
fig.add_trace(go.Bar(
    x=RETRIEVED_METRICS,
    y=score_bge,
    name='bge-m3',
    marker_color='indianred'
))
fig.add_trace(go.Bar(
    x=RETRIEVED_METRICS,
    y=score_jina,
    name='jina-embedding-v3',
    marker_color='lavenderblush'
))
fig.add_trace(go.Bar(
    x=RETRIEVED_METRICS,
    y=score_e5,
    name='multilingual-e5-large-instruct',
    marker_color='lightsalmon'
))
fig.add_trace(go.Bar(
    x=RETRIEVED_METRICS,
    y=score_nomic,
    name='nomic-embed-text-v2-moe',
    marker_color='thistle'
))
fig.add_trace(go.Bar(
    x=RETRIEVED_METRICS,
    y=score_labse,
    name='LaBSE',
    marker_color='lightblue'
))

# Here we modify the tickangle of the xaxis, resulting in rotated labels.
fig.update_layout(title="Embedding model performance", title_x=0.5)
fig.update_layout(barmode='group', xaxis_tickangle=0)
fig.update_yaxes(range=[0, 1], title="score")
fig.show()
```
